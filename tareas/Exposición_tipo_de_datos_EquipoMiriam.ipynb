**EJEMPLO DE DATOS PRIMITIVOS**

utilizaremos el tipo de dato primitivo int para guardar la edad del usuario. El programa solicitará su edad (se colocara dentro de "echo "x""), realizará una suma y mostrará el resultado en pantalla

```python
%%script bash
cat << 'EOF' > edad.c
#include <stdio.h>

int main(void)
{
    int edad;

    printf("¿cuantos años tienes papu? \n");
    scanf("%d", &edad);

    printf("ok papu tu edad ahora es %d años\n", edad);
    printf("El proximo año tendras %d años equisde\n", edad + 1);

    return 0;
}
EOF

gcc edad.c -o edad
# Aquí pasamos la edad (por ejemplo, 18) directamente usando 'echo'
echo "18" | ./edad
```

**EJEMPLO DE DATOS DERIVADOS**

Usaremos un arreglo (array) para guardar la calificaciones de 5 materias por alumno, y al final se mostrará una impresión de pantalla de todas las calificaciones.

```python
%%script bash
cat << 'EOF' >  calificaciones.c

#include <stdio.h>
8
int main()
{
  int i;
  float calificaciones[5];
  for (int i = 0; i < 5; i++) {
        printf("Calificación %d: ", i + 1);
        scanf("%f", &calificaciones[i]);
    }

    printf("\nCalificaciones capturadas:\n");
    for (int i = 0; i < 5; i++) {
        printf("%d. %.1f\n", i + 1, calificaciones[i]);
    }

    return 0;
}
EOF

gcc calificaciones.c -o calificaciones
```
