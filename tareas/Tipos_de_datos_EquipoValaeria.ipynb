# **Datos Derivados**

***1. Arreglos (Arrays / Formaciones)***

¿Qué es? Es un tipo derivado porque construye una secuencia fija y continua en memoria a partir de un tipo de dato base (en este caso, enteros).   
¿Cómo se usa?
Se definen indicando el tipo de dato, el nombre y el tamaño entre corchetes [tamaño]. Se accede a cada elemento mediante un índice entero que inicia en 0.

```python
%%script bash
cat << 'EOF' > ejemplo_arreglos.c
#include <stdio.h>

int main() {
    // TIPO DERIVADO: Arreglo de 5 elementos tipo entero
    int numeros[5] = {10, 20, 30, 40, 50};

    printf("--- EJEMPLO DE ARREGLO ---\n");

    // Recorremos el arreglo usando un bucle 'for'
    for (int i = 0; i < 5; i++) {
        printf("Indice [%d] -> Valor almacenado: %d\n", i, numeros[i]);
    }

    return 0;
}
```

***2. Punteros (Pointers / Apuntadores)***

¿Qué es? Es un tipo derivado porque no almacena un valor numérico o un texto directamente, sino la dirección de memoria física donde está guardada otra variable.   
¿Cómo se usa? Se declara anteponiendo el asterisco * al nombre de la variable (por ejemplo, int *ptr). Para obtener la dirección de una variable se usa el operador &, y para acceder al valor guardado en esa dirección se usa *ptr.

```python
%%script bash
cat << 'EOF' > ejemplo_punteros.c
#include <stdio.h>

int main() {
    int numero = 100;

    // TIPO DERIVADO: Puntero a entero (guarda la dirección de 'numero')
    int *ptr = &numero;

    printf("--- EJEMPLO DE PUNTERO ---\n");
    printf("Valor de la variable original (numero): %d\n", numero);
    printf("Direccion de memoria de numero (&numero): %p\n", (void*)&numero);
    printf("Direccion guardada dentro del puntero (ptr): %p\n", (void*)ptr);
    printf("Valor obtenido desreferenciando el puntero (*ptr): %d\n", *ptr);

    return 0;
}
```

```python
!gcc ejemplo_punteros.c -o ejemplo_punteros && ./ejemplo_punteros
```

***Funciones***

¿Qué es? En C, una función se considera un tipo derivado porque su "tipo" no es un dato único, sino la combinación del tipo de dato que retorna y los tipos de parámetros que recibe.   
¿Cómo se usa? Se declara indicando el tipo de retorno (como int o void), el nombre de la función y entre paréntesis sus parámetros de entrada. Luego se invoca enviándole los valores requeridos.   

```python
%%script bash
cat << 'EOF' > ejemplo_funciones.c
#include <stdio.h>

// TIPO DERIVADO: Función cuyo tipo derivado es (int, int) -> int
int sumar(int a, int b) {
    return a + b; // Retorna la suma de los dos enteros
}

int main() {
    int num1 = 15;
    int num2 = 25;

    printf("--- EJEMPLO DE FUNCIÓN ---\n");

    // Invocamos la función pasándole num1 y num2 como argumentos
    int resultado = sumar(num1, num2);

    printf("El resultado de sumar %d + %d es: %d\n", num1, num2, resultado);

    return 0;
}
```

```python
!gcc ejemplo_funciones.c -o ejemplo_funciones && ./ejemplo_funciones
```

**Datos Primitivos**

Los tipos de datos primitivos (también llamados básicos o fundamentales) son los tipos más simples que vienen definidos por el lenguaje. No se pueden dividir en partes más pequeñas y sirven como base para construir los tipos compuestos.

**Modificadores**

Los modificadores se colocan antes de un tipo de dato básico para cambiar su tamaño o si acepta números negativos.

***Ejemplo 1 sin boleanos***

```python
%%script bash
cat << 'EOF' > ejemplo_primitivos.c
#include <stdio.h>

int main() {
    // Tipos primitivos básicos
    char letra = 'A';
    int edad = 20;
    float altura = 1.75f;      // la "f" es correcta en C
    double pi = 3.1415926535;

    // Con modificadores
    short pequeno = 32000;
    unsigned int positivo = 4000000000U;
    long largo = 2000000000L;
    long long muyLargo = 9000000000000000000LL;
    unsigned char byte = 255;

    printf("=== DATOS PRIMITIVOS ===\n");
    printf("char: %c\n", letra);
    printf("int: %d\n", edad);
    printf("float: %.2f\n", altura);
    printf("double: %.10f\n\n", pi);

    printf("=== CON MODIFICADORES ===\n");
    printf("short: %hd\n", pequeno);
    printf("unsigned int: %u\n", positivo);
    printf("long: %ld\n", largo);
    printf("long long: %lld\n", muyLargo);
    printf("unsigned char: %u\n", byte);

    return 0;
}
EOF
```

```python
!gcc ejemplo_primitivos.c -o ejemplo_primitivos && ./ejemplo_primitivos
```

***Ejemplo 2 Con boleanos (incluye todos los mencionados en la presentación***

Explicación general del programa

Los tipos de datos primitivos son los bloques básicos e indivisibles de
almacenamiento en C (int, float, double, char, bool).   

Los modificadores (short, long, signed, unsigned) son palabras clave que alteran el tamaño en memoria RAM o cambian el rango numérico (con o sin signo) de esos tipos primitivos.  

```python
%%script bash
cat << 'EOF' > ejemplo_primitivos_modificadores.c
#include <stdio.h>
#include <stdatomic.h>
#include <stdbool.h> // Necesario para el tipo primitivo bool (true / false)

int main() {
    printf("==================================================\n");
    printf("         1. TIPOS PRIMITIVOS BÁSICOS             \n");
    printf("==================================================\n");

    // INT: Números enteros (positivos, negativos o cero)
    int entero = 42;

    // FLOAT: Decimales de precisión simple
    float flotante = 3.1416f;

    // DOUBLE: Decimales de doble precisión
    double doble = 3.141592653589793;

    // CHAR: Un solo carácter o símbolo ASCII
    char letra = 'A';

    // BOOL: Tipo booleano (verdadero = 1, falso = 0)
    bool es_valido = true;

    // Impresión de valores y uso de sizeof() para medir su tamaño en memoria
    printf("int:    Valor = %d   | Tamano real = %lu bytes\n", entero, sizeof(entero));
    printf("float:  Valor = %.4f | Tamano real = %lu bytes\n", flotante, sizeof(flotante));
    printf("double: Valor = %.11lf | Tamano real = %lu bytes\n", doble, sizeof(doble));
    printf("char:   Valor = %c    | Tamano real = %lu byte\n", letra, sizeof(letra));
    printf("bool:   Valor = %s  | Tamano real = %lu byte\n\n", es_valido ? "true" : "false", sizeof(es_valido));


    printf("==================================================\n");
    printf("   2. MODIFICADORES DE TAMAÑO Y SIGNO            \n");
    printf("==================================================\n");

    // SHORT: Reduce el espacio reservado a un mínimo de 2 bytes (-32,768 a 32,767)
    short int entero_corto = -32000;

    // LONG LONG: Expande el espacio a 8 bytes para enteros masivos (millones de billones)
    long long int entero_gigante = 9000000000000000000LL;

    // SIGNED: Reserva 1 bit para el signo (+ o -). Es el comportamiento por defecto
    signed int entero_con_signo = -150;

    // UNSIGNED: Elimina el signo positivo/negativo y almacena solo valores >= 0,
    // duplicando la capacidad máxima positiva en el mismo espacio de memoria
    unsigned int entero_sin_signo = 4000000000U;

    // MODIFICADOR EN CHAR: Permite guardar valores numéricos de 0 a 255
    unsigned char caracter_sin_signo = 255;

    printf("short int:     Valor = %d | Tamano real = %lu bytes\n", entero_corto, sizeof(entero_corto));
    printf("long long int: Valor = %lld | Tamano real = %lu bytes\n", entero_gigante, sizeof(entero_gigante));
    printf("signed int:    Valor = %d | Tamano real = %lu bytes\n", entero_con_signo, sizeof(entero_con_signo));
    printf("unsigned int:  Valor = %u | Tamano real = %lu bytes\n", entero_sin_signo, sizeof(entero_sin_signo));
    printf("unsigned char: Valor = %u | Tamano real = %lu byte\n", caracter_sin_signo, sizeof(caracter_sin_signo));

    return 0;
}
```

#   **Datos Estructurados o compuestos**



***1. Estructuras (`struct`)***



**Agrupación Heterogénea**

Permiten juntar variables de distintos tipos bajo un mismo nombre. Es algo así como un dato hecho de diferentes tipos de variables

```python
%%script bash
cat << 'EOF' > for_basico.c

#include <stdio.h>


struct Estudiante {
    char nombre[30];
    int edad;
    float promedio;
};

int main() {
    struct Estudiante e1 = {"Carlos", 20, 9.4f};
    printf("Alumno: %s, Edad: %d, Promedio: %.2f\n", e1.nombre, e1.edad, e1.promedio);
    return 0;
}
EOF
gcc for_basico.c -o for_basico
./for_basico
```

Usar datos estructurados  (`struct`) en lugar de variables separadas se vuelve indispensable en cuanto tu programa empieza a manejar más de un elemento o información que está relacionada entre sí.

Pues imaginemos que tenemos que enlistar a muchos estudiantes, si quisieramos alamacenar cada dato de cada estudiante sin usar  (`struct`) serian varias lineas de codigo declaroando cada variable de cada estudiante




```python
%%script bash
cat << 'EOF' > sin_struct.c

#include <stdio.h>

int main() {
    // Variables separadas para el Estudiante 1
    char nombre1[30] = "Ana";
    int edad1 = 19;
    float promedio1 = 9.5f;

    // Variables separadas para el Estudiante 2
    char nombre2[30] = "Carlos";
    int edad2 = 20;
    float promedio2 = 8.8f;

    // Variables separadas para el Estudiante 3
    char nombre3[30] = "Sofia";
    int edad3 = 21;
    float promedio3 = 9.2f;

    printf("=== LISTA DE ESTUDIANTES (SIN Struct - Variables Separadas) ===\n\n");

    // Imprimir requiere escribir una línea distinta para cada persona
    // (no se puede usar un ciclo for porque las variables no están indexadas)
    printf("Estudiante 1 -> Nombre: %s | Edad: %d | Promedio: %.2f\n", nombre1, edad1, promedio1);
    printf("Estudiante 2 -> Nombre: %s | Edad: %d | Promedio: %.2f\n", nombre2, edad2, promedio2);
    printf("Estudiante 3 -> Nombre: %s | Edad: %d | Promedio: %.2f\n", nombre3, edad3, promedio3);

    return 0;
}
EOF

gcc sin_struct.c -o sin_struct
./sin_struct
```

Cero repetición: En lugar de crear 3 variables diferentes (nombre1, nombre2, nombre3, edad1...), solo creas un tipo de dato y un arreglo.

```python
%%script bash
cat << 'EOF' > ejemplo_struct.c

#include <stdio.h>

// Definimos la estructura del estudiante

struct Estudiante {
    char nombre[30];
    int edad;
    float promedio;
};

int main() {
    // Creamos un arreglo de 3 estudiantes con sus datos ya precargados

    struct Estudiante grupo[3] = {
        {"Ana", 19, 9.5f},
        {"Carlos", 20, 8.8f},
        {"Sofia", 21, 9.2f}
    };
    //Como ven ya no necesitamos declarar que tipo de variables son cada dato

    printf("LISTA DE ESTUDIANTES\n\n");

    // Imprimimos todo de forma limpia usando un ciclo
    for (int i = 0; i < 3; i++) {
        printf("Estudiante %d -> Nombre: %s | Edad: %d | Promedio: %.2f\n",  i + 1, grupo[i].nombre, grupo[i].edad, grupo[i].promedio);
    }

    return 0;
}
EOF

gcc ejemplo_struct.c -o ejemplo_struct
./ejemplo_struct
```

***2. Enumeraciones (enum)***



¿Qué es? Es un tipo compuesto que permite asignar nombres simbólicos y descriptivos a un conjunto de constantes enteras (0, 1, 2...).

Sirven para darle nombres claros y descriptivos a números enteros."


Imaginen que en lugar de recordar que el número 1 significa 'Lunes', el 2 'Martes' y así sucesivamente, creamos un enum para usar directamente las palabras LUNES, MARTES, etc. Esto hace que nuestro código sea mucho más fácil de leer y menos propenso a errores."

```python
%%writefile dias.c
#include <stdio.h>

// Definimos el enum para los días de la semana
enum DiaSemana {
    LUNES,      // 0
    MARTES,     // 1
    MIERCOLES,  // 2
    JUEVES,     // 3
    VIERNES,    // 4
    SABADO,     // 5
    DOMINGO     // 6
};

int main() {
    // Declaramos una variable de tipo enum DiaSemana
    enum DiaSemana hoy = MIERCOLES;

    printf("El valor numerico interno del dia es: %d\n", hoy);

    // Evaluamos la variable con switch
    switch(hoy) {
        case LUNES:
            printf("¡Animo, es Lunes! Comienza la semana.\n");
            break;
        case MARTES:
            printf("Es Martes, seguimos avanzando.\n");
            break;
        case MIERCOLES:
            printf("¡Es Miercoles! Ya estamos a mitad de semana.\n");
            break;
        case JUEVES:
            printf("Es Jueves, casi es viernes.\n");
            break;
        case VIERNES:
            printf("¡Por fin es Viernes!\n");
            break;
        case SABADO:
        case DOMINGO:
            printf("¡Es fin de semana, a descansar!\n");
            break;
        default:
            printf("Dia no valido.\n");
    }

    return 0;
}
```

```python

!gcc dias.c -o dias && ./dias
```

***3. Uniones (union)***

¿Qué es? Es una agrupación similar a la estructura, pero con una gran diferencia: todos sus miembros comparten exactamente la misma posición de memoria RAM.

¿Para qué sirve? Sirve para ahorrar espacio de memoria en sistemas donde una variable puede ser de un tipo u otro, pero nunca de ambos tipos al mismo tiempo. Su tamaño en bytes equivale al de su miembro más grande.

```python
%%script bash
cat << 'EOF' > ejemplo_union.c
#include <stdio.h>

// TIPO COMPUESTO: Unión donde el dato puede ser entero o flotante
union Dato {
    int entero;
    float flotante;
};

int main() {
    union Dato mi_dato;

    printf("--- EJEMPLO DE UNION ---\n");
    printf("Tamano ocupado por la union: %lu bytes\n\n", sizeof(union Dato));

    // 1. Guardamos un valor entero
    mi_dato.entero = 42;
    printf("Valor guardado como entero: %d\n", mi_dato.entero);

    // 2. Guardamos un flotante (sobrescribe el espacio de memoria anterior)
    mi_dato.flotante = 3.1416;
    printf("Valor guardado como flotante: %.4f\n", mi_dato.flotante);

    // Si leemos el entero de nuevo, estará alterado porque la memoria fue sobrescrita
    printf("Entero tras guardar flotante (dato alterado): %d\n", mi_dato.entero);

    return 0;
}
```

```python
!gcc ejemplo_union.c -o ejemplo_union && ./ejemplo_union
```
