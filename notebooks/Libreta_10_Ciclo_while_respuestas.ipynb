# Ciclo While

La estructura while en el lenguaje de programación C es un bucle que repite un bloque de código mientras una condición dada sea verdadera (distinta de cero)

```
Sintáxis
while (condición) {
    // Instrucciones que se repiten
}
```

**condición:** Se evalúa antes de cada iteración. Si al empezar es falsa (0), el bloque de código no se ejecuta ni una sola vez.

Recuerda: La condición siempre se revisa antes de entrar al ciclo

## Ejercicio 1: Números naturales
Imprime los números del 1 al 5 usando un contador


````python
%%script bash
cat << 'EOF' > ejercicio1_while.c

#include <stdio.h> //Header

int main() {

  //declaración de variables
  int i = 1; //contador

  // Inicio del ciclo while
  while (i <= 5) // Condición
  {
    printf("Numero: %d\n", i);
    i++; // Actualización del contador
  }
    return 0;
}

EOF
gcc ejercicio1_while.c -o ejercicio1_while
````

## Ejercicio 4: Validación de password

````python
%%script bash
cat << 'EOF' > ejercicio4_while.c

#include <stdio.h> //Header
int main()
{
  //Declaración de variables
  int pin = 123;
  int input;

  //inicio del ciclo while
  while (1)
  {
    printf("Ingresa el password: ");
    scanf("%d", &input);

    if (input == pin)
    {
      printf("Password correcto\n");
      break;  // Salir del ciclo
    }
    else
    {
      printf("Password incorrecto, intenta otra vez.\n");
    }
  }
  return 0;
}

EOF
gcc ejercicio4_while.c -o ejercicio4_while
````

Modifica el código anterior, para que compare cadenas, en lugar de números.

````python
%%script bash
cat << 'EOF' > ejercicio7_while.c

#include <stdio.h> //Header

int main() {

    //declaración de variables
    float monto_compra = 0.0;
    float total_ventas = 0.0;
    int contador_clientes = 0;
    char continuar = 's';

    printf("=== SISTEMA DE REGISTRO DE VENTAS ===\n");

    // Ciclo while para registrar ventas mientras el usuario desee continuar

    while (continuar == 's' || continuar == 'S') {
        printf("\nIngrese el monto de la compra del cliente: $");
        scanf("%f", &monto_compra);

        // Acumulamos el monto y contamos al cliente
        total_ventas += monto_compra;
        contador_clientes++;

        printf("¿Desea registrar la compra de otro cliente? (s/n): ");
        scanf(" %c", &continuar); // El espacio antes de %c limpia el buffer del salto de línea
    }

    // Resultados del día
    printf("\n=====================================\n");
    printf("         RESUMEN DEL DÍA             \n");
    printf("=====================================\n");
    printf("Número total de clientes atendidos: %d\n", contador_clientes);
    printf("Monto total de ventas: $%.2f\n", total_ventas);
    printf("=====================================\n");

    return 0;
}

EOF
gcc ejercicio7_while.c -o ejercicio7_while
````
