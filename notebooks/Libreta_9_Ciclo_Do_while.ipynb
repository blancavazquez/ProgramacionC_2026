## Ciclo Do-while

El objetivo de esta libreta es practicar la estructura de control Do-while.


* **do-while** en C es una estructura de control que ejecuta un bloque de código al menos una vez, y luego evalúa una condición para decidir si el bloque se repite.

```
Sintáxis
do {
    // Bloque de instrucciones que se ejecuta al menos una vez
} while (condición);
```

## Ejercicio 1: Suma de números hasta ingresar un cero

Este programa le pide al usuario que escriba números enteros y los va sumando. El ciclo se repite hasta que el usuario escriba el número 0

````python
%%script bash
cat << 'EOF' > ejercicio1_dowhile.c

#include <stdio.h> //Header

int main() {

    //declaración de variables
    int numero;
    int suma = 0;

    // Inicio del ciclo Do
    do {
        printf("Ingresa un número (0 para terminar): ");
        scanf("%d", &numero);
        suma += numero; // Acumula el valor en la variable suma

    // cierre del ciclo
    }
    while (numero != 0); // Se repite mientras el número sea diferente de 0

    printf("\nLa suma total de los números es: %d\n", suma);

    return 0;
}


EOF
gcc ejercicio1_dowhile.c -o ejercicio1_dowhile
````

````python
!gcc ejercicio1_dowhile.c -o ejercicio1_dowhile -lm && ./ejercicio1_dowhile
````

# Ejercicio 2: Validación de password


Escribe un programa en C que valide el password ingresado por el usuario

````python
%%script bash
cat << 'EOF' > ejercicio2_dowhile_passsword.c

#include <stdio.h> //Header
#include <string.h> //Header

int main() {

    //declaración de variables
    char correctPassword[] = "pass123";  // Escribe el password correcto
    char enteredPassword[20];  // Variable para almacenar el password que ingresa el usuario

    // Inicio del ciclo Do
    do
    {
        printf("Ingresa el password: ");
        scanf("%s", enteredPassword);

        // Verifica si el password es correcto
        if (strcmp(enteredPassword, correctPassword) == 0)
        {
          printf("Password correcto.\n");
          break;
        }
        else
        {
          printf("Password incorrecto. Intenta otra vez.\n");
        }

    //Cierre del ciclo
    }
    while (1);  // El ciclo se ejecutará hasta ¿?

    printf("****************************");

    return 0;  // Cierre de ejecución del programa
}

EOF
gcc ejercicio2_dowhile_passsword.c -o ejercicio2_dowhile_passsword
````
