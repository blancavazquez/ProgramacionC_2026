**Ciclo for:**

Es una estructura de control repetitiva que permite ejecutar un bloque de código un número determinado de veces. Su sintáxtis es:

for (inicialización; condición; incremento)

{

  Código que se ejecuta en cada repetición

}





## Ejercicio 1: Imprimir los primeros 10 números naturales

Este ejercicio muestra cómo usar un contador básico que inicia en 1 y se incrementa en cada paso hasta llegar a 10

```python
%%script bash
cat << 'EOF' > for_basico.c

#include <stdio.h> // cabecera para realizar operaciones de entrada y salida estándar
int main() // es la función principal y el punto de entrada de un programa
{
    int i;
    for (i = 1; i <= 10; i++)
    {
      printf("%d ", i);
    }
    printf("\n");
    return 0;
}

EOF
gcc for_basico.c -o for_basico
./for_basico
```
