## Operaciones simples

El objetivo de esta libreta es interactuar con datos ingresados desde teclado y realizar operaciones de cálculo de aŕea.

### Rectángulo

```python
%%writefile rectangle.c
#include <stdio.h>

int main() {
    float length, width, area;

    printf("Indica el largo del rectángulo: ");
    scanf("%f", &length);

    printf("Indica el ancho del rectángulo: ");
    scanf("%f", &width);

    area = length * width;

    printf("El área del rectángulo: %.2f\n", area);

    return 0;
}
```

```python
!gcc rectangle.c -o rectangle
```

```python
!gcc rectangle.c -o rectangle -lm && ./rectangle
```

### Rectángulo (versión 2)

```python
%%writefile rectangle_function.c
#include <stdio.h>

float calculateArea(float length, float width)
{
    return length * width;
}

int main() {
    float length, width, area;


    printf("Indica el largo del rectángulo: ");
    scanf("%f", &length);

    printf("Indica el ancho del rectángulo: ");
    scanf("%f", &width);

    area = calculateArea(length, width);

    printf("El área del rectángulo: %.2f\n", area);

    return 0;
}
```

### Círculo

```python
%%writefile calculate_area.c
#include <stdio.h>
#include <math.h>

int main() {
    float radius, area;

    printf("Indica el radio del círculo: ");
    if (scanf("%f", &radius) == 1)
    {
        area = M_PI * pow(radius, 2); //Calculando el área usando la función pi (definida en math.h)
        printf("El área de un círculo con radio %.2f is: %.2f\n", radius, area);
    }
    else
    {
        printf("Entrada invalida.\n");
    }
    return 0;
}
```
