## Estructura Switch-case

La estructura switch en C es una estructura de control condicional múltiple que permite evaluar el valor de una variable o expresión y compararlo con distintos casos predefinidos

```
Sintáxis
switch (expresion) {
    case valor1:
        // Instrucciones a ejecutar si expresion == valor1
        break;
    case valor2:
        // Instrucciones a ejecutar si expresion == valor2
        break;
    default:
        // Instrucciones si no coincide con ningún caso (opcional)
}
```

La expresión que evalúa el switch solo puede ser de tipo entero (int) o carácter (char). No acepta números decimales (float, double) ni cadenas de texto directamente

# Ejercicio 1: Calculadora básica

````python
%%script bash
cat << 'EOF' > ejercicio1_case.c
#include <stdio.h>

int main() {
    char operador;
    double num1, num2;

    printf("Introduce un operador (+, -, *, /): ");
    scanf("%c", &operador);

    printf("Introduce dos operandos: ");
    scanf("%lf %lf", &num1, &num2);

    switch (operador) {
        case '+':
            printf("%.2lf + %.2lf = %.2lf\n", num1, num2, num1 + num2);
            break;
        case '-':
            printf("%.2lf - %.2lf = %.2lf\n", num1, num2, num1 - num2);
            break;
        case '*':
            printf("%.2lf * %.2lf = %.2lf\n", num1, num2, num1 * num2);
            break;
        case '/':
            if (num2 != 0)
                printf("%.2lf / %.2lf = %.2lf\n", num1, num2, num1 / num2);
            else
                printf("Error: División por cero.\n");
            break;
        default:
            printf("Error: Operador no válido.\n");
            break;
    }

    return 0;
}

EOF
gcc ejercicio1_case.c -o ejercicio1_case
````

## Ejercicio 3: Estructuras

````python
%%script bash
cat << 'EOF' > ejercicio_estructuras.c
#include <stdio.h>

// 1. Definir la estructura
struct Persona {
    char nombre[50];
    int edad;
    float altura;
};

int main() {
    // 2. Declarar una variable de tipo estructura
    struct Persona p1;

    // 3. Capturar datos desde el teclado
    printf("Ingrese el nombre: ");
    scanf("%s", p1.nombre); // Lee una cadena sin espacios

    printf("Ingrese la edad: ");
    scanf("%d", &p1.edad); // Usa & para pasar la dirección de memoria

    printf("Ingrese la altura (en metros): ");
    scanf("%f", &p1.altura);

    // 4. Imprimir los datos capturados
    printf("\n--- Datos Guardados ---\n");
    printf("Nombre: %s\n", p1.nombre);
    printf("Edad: %d años\n", p1.edad);
    printf("Altura: %.2f m\n", p1.altura);

    return 0;
}


EOF
gcc ejercicio_estructuras.c -o ejercicio_estructuras
````

````python
!gcc ejercicio_estructuras.c -o ejercicio_estructuras -lm && ./ejercicio_estructuras
````
