```python
%%script bash
cat << 'EOF' > tarea_switch_case.c

#include <stdio.h>
#include <string.h>

#define TOTAL_PRODUCTOS 4
#define MAX_LETRA 50

int main() {
    // Inventario del sistema
    char nombres[TOTAL_PRODUCTOS][MAX_LETRA] = {"Gorra", "Camiseta", "Pantalon", "Zapatos"};
    float precios[TOTAL_PRODUCTOS] = {15.50, 20.00, 35.99, 59.99};
    int stock[TOTAL_PRODUCTOS] = {10, 15, 8, 5};

    int opcion;
    char producto_buscar[MAX_LETRA];
    int cantidad_venta;
    int encontrado;
    float total_venta = 0.0;

    do {
        // Menú Principal
        printf("\n--- SISTEMA DE VENTAS ---\n");
        printf("1. Ver inventario disponible\n");
        printf("2. Realizar una venta\n");
        printf("3. Ver total de ventas del dia\n");
        printf("4. Salir\n");
        printf("Seleccione una opcion: ");
        scanf("%d", &opcion);
        getchar(); // Limpiar el buffer del salto de línea

        switch (opcion) {
            case 1:
                printf("\n--- INVENTARIO ACTUAL ---\n");
                printf("%-15s %-10s %-10s\n", "Producto", "Precio", "Stock");
                // Uso de FOR para listar los productos
                for (int i = 0; i < TOTAL_PRODUCTOS; i++) {
                    printf("%-15s $%-9.2f %-10d\n", nombres[i], precios[i], stock[i]);
                }
                break;

            case 2:
                printf("\n--- REALIZAR VENTA ---\n");
                printf("Ingrese el nombre del producto: ");
                fgets(producto_buscar, MAX_LETRA, stdin);
                producto_buscar[strcspn(producto_buscar, "\n")] = 0; // Remover el salto de línea de fgets

                encontrado = 0;

                // Uso de FOR para buscar el producto solicitado
                for (int i = 0; i < TOTAL_PRODUCTOS; i++) {
                    // Uso de STRCMP para comparar cadenas de texto (retorna 0 si son iguales)
                    if (strcmp(nombres[i], producto_buscar) == 0) {
                        encontrado = 1;
                        printf("Producto encontrado: %s ($%.2f) - Disponibles: %d\n", nombres[i], precios[i], stock[i]);
                        printf("Ingrese la cantidad a vender: ");
                        scanf("%d", &cantidad_venta);

                        // Uso de IF para verificar disponibilidad de existencias
                        if (cantidad_venta <= stock[i] && cantidad_venta > 0) {
                            float subtotal = precios[i] * cantidad_venta;
                            stock[i] -= cantidad_venta; // Restar del inventario
                            total_venta += subtotal;    // Acumular en la caja del día
                            printf("¡Venta exitosa! Subtotal: $%.2f\n", subtotal);
                        } else {
                            printf("Error: No hay suficiente stock o cantidad invalida.\n");
                        }
                        break; // Rompe el ciclo FOR porque ya encontró el producto
                    }
                }

                // Uso de IF por si el artículo no existe en la base de datos
                if (!encontrado) {
                    printf("El producto '%s' no existe en el inventario.\n", producto_buscar);
                }
                break;

            case 3:
                printf("\n--- CORTE DE CAJA ---\n");
                printf("El total de ventas acumulado hoy es: $%.2f\n", total_venta);
                break;

            case 4:
                printf("\nSaliendo del sistema... ¡Que tenga un buen dia!\n");
                break;

            default:
                printf("\nOpcion no valida. Intente de nuevo.\n");
        }

    } while (opcion != 4);

    return 0;
}


EOF
gcc tarea_switch_case.c -o tarea_switch_case
```
