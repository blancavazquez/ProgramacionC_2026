## Código ASCII


ASCII son las siglas de **American Standard Code for Information Interchange** (Código Estándar Estadounidense para el Intercambio de Información).

* Se trata de un sistema de codificación que asigna un valor numérico único a cada carácter, letra, número o símbolo utilizado en la comunicación digital.

* Originalmente, el código ASCII utiliza 7 bits para representar hasta 128 caracteres diferentes. Estos incluyen letras mayúsculas y minúsculas del alfabeto inglés (A-Z, a-z), números (0-9), signos de puntuación (., ;, !), y caracteres de control como “Enter” o “Escape”.

* Por ejemplo, la letra “A” mayúscula se representa con el número 65, mientras que la “a” minúscula corresponde al 97.

* Este sistema permite que las computadoras, que solo entienden números, puedan procesar y mostrar texto de manera comprensible para los humanos.

## Ejercicio 1: Contando el número de caracteres de una palabra

Solicita una palabra a un usuario y devuelve el tamaño de la palabra (conteo de caracteres).

```python
%%script bash
cat << 'EOF' > for_cadenas.c

#include <stdio.h>  // Cabecera para entradas / salidas
#include <string.h> // Cabecera para procesar cadenas

int main()
{
  char cadena1 [100];
  char cadena2 [100];

  printf("\nIngresa la cadena 1: ");

  //Evaluando las siguientes líneas de código
  //scanf("%s", cadena1);
  fgets(cadena1, sizeof(cadena1), stdin); // stdin = standard input

  int length = strlen(cadena1); // encontrar el tamaño de la cadena
  printf("El tamaño de la cadena: %s es %d", cadena1, length);
  return 0;
}

EOF
gcc for_cadenas.c -o for_cadenas
```

##Ejercicio 2: Comparación entre cadenas

Solicita al usuario dos cadenas de texto, devuelve cuál de ellas es mayor.


```python
%%script bash
cat << 'EOF' > for_comparativo.c

#include <stdio.h>  // Cabecera para entradas / salidas
#include <string.h> // Cabecera para procesar cadenas

void main()
{
  char cadena1[100];
  char cadena2[100];

  printf("\nIngresa la cadena 1: ");
  fgets(cadena1, sizeof(cadena1), stdin); // Lee la cadena

  printf("\nIngresa la cadena 2: ");
  fgets(cadena2, sizeof(cadena2), stdin); //

  int resultado = strcmp(cadena1, cadena2); //sirve para comparar dos cadenas de caracteres (strings) de forma lexicográfica

  printf("****************\n");

  if (resultado == 0)
  {
    printf("Las cadenas son iguales:\n %s %s", cadena1, cadena2);
  }
  else if (resultado < 0) //valor negativo
  {
    printf("La cadena1 es menor que la cadena2:\n %s %s", cadena1, cadena2);
  }
  else // valor mayor
  {
    printf("La cadena1 es mayor que cadena2:\n %s %s", cadena1, cadena2);
  }

  printf("\nresultado: %d\n", resultado);
}

EOF
gcc for_comparativo.c -o for_comparativo
```

```python
!gcc for_comparativo.c -o for_comparativo -lm && ./for_comparativo
```

## ¿Cómo trabaja strcmp?

La función strcmp() (perteneciente a la librería <string.h> en C y C++) compara dos cadenas de caracteres carácter por carácter utilizando sus valores en la tabla ASCII.

Devuelve:
* 0 cuando las cadenas son exactamente iguales.
* Menor que 0 (negativo): La primera cadena (cadena1) es alfabéticamente menor que la segunda (cadena2), o el primer carácter diferente tiene un valor ASCII menor.
* Mayor que 0 (positivo): La primera cadena (cadena1) es alfabéticamente mayor que la segunda (cadena2), o el primer carácter diferente tiene un valor ASCII mayor.
