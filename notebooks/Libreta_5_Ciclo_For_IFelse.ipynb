# Libreta: Ciclo For

El objetivo de esta libreta es practicar el ciclo For y la instrucció if-else

## Ejercicio 1: Calcular promedios

```python
%%script bash
cat << 'EOF' > for_promedios.c

#include <stdio.h>

int main() {
    int num_alumnos;
    float calificacion;
    float NOTA_MINIMA = 60.0; // Definimos la nota mínima para aprobar

    printf("=== SISTEMA DE CALIFICACIONES ===\n\n");
    printf("Ingrese la cantidad de estudiantes a registrar en el sistema: "); // Solicitar cuántos alumnos se van a evaluar
    scanf("%d", &num_alumnos);

    if (num_alumnos <= 0) { // Validar que la cantidad sea un número positivo
        printf("La cantidad de alumnos debe ser mayor a cero.\n");
        return 1;
    }

    for (int i = 1; i <= num_alumnos; i++) { // Bucle para procesar a cada alumno
        printf("\n--- Alumno %d ---\n", i);
        printf("Ingrese la calificación (0 a 100): ");
        scanf("%f", &calificacion);

        if (calificacion < 0 || calificacion > 100) { // Validar que la calificación esté dentro de un rango lógico
            printf("Calificación inválida. Por favor, intente de nuevo con este estudiante.\n");
            i--; // Decrementamos para repetir el ingreso del alumno actual
            continue;
        }

        if (calificacion >= NOTA_MINIMA) // Evaluar el estatus utilizando una condicional if-else
        {
            printf("Estado: APROBADO\n");
        } else
        {
            printf("Estado: REPROBADO\n");
        }
    }

    printf("\nProceso finalizado correctamente.\n");
    return 0;
}

EOF
gcc for_promedios.c -o for_promedios
```
